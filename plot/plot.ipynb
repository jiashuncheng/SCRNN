```python
import pickle
import numpy as np
from sklearn.decomposition import PCA
from sklearn.manifold import Isomap
import matplotlib.pyplot as plt
from dPCA import dPCA
import math

class Bezier:
    # 输入控制点，Points是一个array,num是控制点间的插补个数
    def __init__(self,Points,InterpolationNum):
        self.demension=Points.shape[1]   # 点的维度
        self.order=Points.shape[0]-1     # 贝塞尔阶数=控制点个数-1
        self.num=InterpolationNum        # 相邻控制点的插补个数
        self.pointsNum=Points.shape[0]   # 控制点的个数
        self.Points=Points
        
    # 获取Bezeir所有插补点
    def getBezierPoints(self,method):
        if method==0:
            return self.DigitalAlgo()
        if method==1:
            return self.DeCasteljauAlgo()
    
    # 数值解法
    def DigitalAlgo(self):
        PB=np.zeros((self.pointsNum,self.demension)) # 求和前各项
        pis =[]                                      # 插补点
        for u in np.arange(0,1+1/self.num,1/self.num):
            for i in range(0,self.pointsNum):
                PB[i]=(math.factorial(self.order)/(math.factorial(i)*math.factorial(self.order-i)))*(u**i)*(1-u)**(self.order-i)*self.Points[i]
            pi=sum(PB).tolist()                      #求和得到一个插补点
            pis.append(pi)            
        return np.array(pis)
def normalization1(data):
    data = data - np.min(data, axis=0)
    data = data/(np.max(np.abs(data), axis=0)[np.newaxis, :]+0.00001)
    return data

with open('/home/jiashuncheng/code/MANN/plot/data/rsc_h_norm_witchout_b_2.pkl', 'rb') as a:
    x = pickle.load(a)
    target = pickle.load(a)
target = np.argmax(target, axis=1)
Y = np.stack(x).squeeze(2)[:,:,:]
Y = normalization1(Y)
P_data = Y[:, np.where(target==1)[0], :][:,:40,:]
N_data = Y[:, np.where(target==0)[0], :][:,:40,:]
N,T,S = 10,135,2
noise, n_samples = 0.2, 40
data = np.zeros((n_samples,N,S,T))
data[:,:,0,:] = P_data.transpose(1,2,0)
data[:,:,1,:] = N_data.transpose(1,2,0)
R = np.mean(data,0)
R -= np.mean(R.reshape((N,-1)),1)[:,None,None]
dpca = dPCA.dPCA(labels='st',n_components=2,regularizer='auto',join =[{'s': ['s', 'st']}, {'t'}])
dpca.protect = ['t']
print(R.shape, data.shape)
Z = dpca.fit_transform(R,data)
ZZ = Z['st'] # n_components, data[2], time

ZZ0 = Bezier(ZZ[:,0,:].T, 1800)
ZZ0 = ZZ0.getBezierPoints(0).T
ZZ1 = Bezier(ZZ[:,1,:].T, 1800)
ZZ1 = ZZ1.getBezierPoints(0).T
ratio = 1800/135

y = ZZ0[:,:int(20*ratio)+1]
plt.scatter(y[0,:], y[1,:], marker='.', color='C2', label='Sample', zorder=2)
for i in range(y.shape[1]-1):
    plt.plot(y[0,i:i+2], y[1,i:i+2], color='C2', linewidth=2)
y = ZZ0[:,int(20*ratio):int(120*ratio)+1]
plt.scatter(y[0,:], y[1,:], marker='.', color='C0', label='Delay', zorder=1)
for i in range(y.shape[1]-1):
    plt.plot(y[0,i:i+2], y[1,i:i+2], color='C0', linewidth=2)
y = ZZ0[:,int(120*ratio):int(136*ratio)]
plt.scatter(y[0,:], y[1,:], marker='.', color='C3', label='Decision', zorder=2)
for i in range(y.shape[1]-1):
    plt.plot(y[0,i:i+2], y[1,i:i+2], color='C3', linewidth=2)

y = ZZ1[:,:int(20*ratio)+1]
plt.scatter(y[0,:], y[1,:], marker='.', color='C5', label='Sample', zorder=2)
for i in range(y.shape[1]-1):
    plt.plot(y[0,i:i+2], y[1,i:i+2], color='C5', linewidth=2)
y = ZZ1[:,int(20*ratio):int(120*ratio)+1]
plt.scatter(y[0,:], y[1,:], marker='.', color='C1', label='Delay', zorder=1)
for i in range(y.shape[1]-1):
    plt.plot(y[0,i:i+2], y[1,i:i+2], color='C1', linewidth=2)
y = ZZ1[:,int(120*ratio):int(136*ratio)]
plt.scatter(y[0,0], y[1,0], marker='.', color='C4', label='Decision', zorder=2)
for i in range(y.shape[1]-1):
    plt.plot(y[0,i:i+2], y[1,i:i+2], color='C4', linewidth=2)
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.cluster.hierarchy import fcluster
import pickle
import matplotlib.pyplot as plt
import torch

def hierarchy_cluster(bb):
    methods_x = [np.argsort(-bb.mean(1))]
    methods_y = [np.argsort(-bb.mean(0))]
    bb=bb[np.argsort(-bb.mean(1)),:]
    # lis = [2,1,4,7,9,8,5,6,0,3]
    lis = [3,0,4,7,8,5,6,1,2,9]
    bb=bb[np.array(lis),:]
    methods_x.append(np.array(lis))
    bb=bb[:,np.argsort(-bb.mean(0))]
    lis = [1,2,3,0,4,5,6,7,8,9]
    bb=bb[:,np.array(lis)]
    methods_y.append(np.array(lis))
    return bb.copy(), methods_x, methods_y

def trans(a, method_x, method_y):
    for i in range(len(method_x)):
        a = a[method_x[i], :]
    for j in range(len(method_y)):
        a = a[:,method_y[j]]
    return a.copy()

# 生成随机数据
with open('/home/jiashuncheng/code/MANN/plot/data/A_seed25.pkl', 'rb') as a:
    b = pickle.load(a)
bb = b[-1,2,:,:]

c, method_x, method_y = hierarchy_cluster(bb)
fig, ax = plt.subplots()
plt.imshow(c)
ax.invert_yaxis()
plt.show()

for i in range(19,20):
    c = b[i,0,:,:]
    c = trans(c, method_x, method_y)
    fig, ax = plt.subplots()
    plt.imshow(c)#, vmax=4, vmin=0)

    import numpy as np
    import pandas as pd
    df = pd.DataFrame(c)
    df.to_excel('g1.xlsx', index=False)

    ax.invert_yaxis()
    cbar = plt.colorbar()
    cbar.ax.tick_params(labelsize=14)  # 设置字体大小为14
    plt.title(f'Left preference memory', fontsize=16, loc='center')
    plt.ylabel('ATN neurons', fontsize=16)
    plt.xlabel('RSC neurons', fontsize=16)
    plt.xticks(range(0,10,2),range(1,11,2),fontsize=16)
    plt.yticks(range(0,10,2),range(1,11,2),fontsize=16)
    # plt.savefig(f'data/figure_tar0/{i+1}.jpg')
    plt.savefig('c1.svg')
    plt.show()
    plt.close()

for i in range(19,20):
    c = b[i,3,:,:]
    c = trans(c, method_x, method_y)
    fig, ax = plt.subplots()
    
    import numpy as np
    import pandas as pd
    df = pd.DataFrame(c)
    df.to_excel('g2.xlsx', index=False)

    plt.imshow(c)#, vmax=4, vmin=0)
    ax.invert_yaxis()
    cbar = plt.colorbar()
    cbar.ax.tick_params(labelsize=14)  # 设置字体大小为14
    plt.title(f'Right preference memory', fontsize=16, loc='center')
    plt.ylabel('ATN neurons', fontsize=16)
    plt.xlabel('RSC neurons', fontsize=16)
    plt.xticks(range(0,10,2),range(1,11,2),fontsize=16)
    plt.yticks(range(0,10,2),range(1,11,2),fontsize=16)
    # plt.savefig(f'data/figure_tar1/{i+1}.jpg')
    plt.savefig('c2.svg')
    plt.show()
    plt.close()
```

```python
import pickle
import numpy as np
from sklearn.decomposition import PCA
from sklearn.manifold import Isomap
import matplotlib.pyplot as plt
from dPCA import dPCA
import math

class Bezier:
    # 输入控制点，Points是一个array,num是控制点间的插补个数
    def __init__(self,Points,InterpolationNum):
        self.demension=Points.shape[1]   # 点的维度
        self.order=Points.shape[0]-1     # 贝塞尔阶数=控制点个数-1
        self.num=InterpolationNum        # 相邻控制点的插补个数
        self.pointsNum=Points.shape[0]   # 控制点的个数
        self.Points=Points
        
    # 获取Bezeir所有插补点
    def getBezierPoints(self,method):
        if method==0:
            return self.DigitalAlgo()
        if method==1:
            return self.DeCasteljauAlgo()
    
    # 数值解法
    def DigitalAlgo(self):
        PB=np.zeros((self.pointsNum,self.demension)) # 求和前各项
        pis =[]                                      # 插补点
        for u in np.arange(0,1+1/self.num,1/self.num):
            for i in range(0,self.pointsNum):
                PB[i]=(math.factorial(self.order)/(math.factorial(i)*math.factorial(self.order-i)))*(u**i)*(1-u)**(self.order-i)*self.Points[i]
            pi=sum(PB).tolist()                      #求和得到一个插补点
            pis.append(pi)            
        return np.array(pis)
def normalization1(data):
    data = data - np.min(data, axis=0)
    data = data/(np.max(np.abs(data), axis=0)[np.newaxis, :]+0.00001)
    return data

with open('/home/jiashuncheng/code/MANN/plot/data/rsc_h_norm_witchout_b_2.pkl', 'rb') as a:
    x = pickle.load(a)
    target = pickle.load(a)
target = np.argmax(target, axis=1)
Y = np.stack(x).squeeze(2)[:,:,:]
Y = normalization1(Y)
P_data = Y[:, np.where(target==1)[0], :][:,:40,:]
N_data = Y[:, np.where(target==0)[0], :][:,:40,:]
N,T,S = 10,135,2
noise, n_samples = 0.2, 40
data = np.zeros((n_samples,N,S,T))
data[:,:,0,:] = P_data.transpose(1,2,0)
data[:,:,1,:] = N_data.transpose(1,2,0)
R = np.mean(data,0)
R -= np.mean(R.reshape((N,-1)),1)[:,None,None]
dpca = dPCA.dPCA(labels='st',n_components=3,regularizer='auto',join =[{'s': ['s', 'st']}, {'t'}])
dpca.protect = ['t']
print(R.shape, data.shape)
Z = dpca.fit_transform(R,data)
ZZ = Z['st'] # n_components, data[2], time

ZZ0 = Bezier(ZZ[:,0,:].T, 1800)
ZZ0 = ZZ0.getBezierPoints(0).T
ZZ1 = Bezier(ZZ[:,1,:].T, 1800)
ZZ1 = ZZ1.getBezierPoints(0).T
ratio = 1800/135

fig=plt.figure()
ax = fig.add_subplot(projection='3d')
y = ZZ0[:,:int(20*ratio)+1]
ax.scatter(y[0,:], y[1,:], y[2,:], marker='.', color='C2', label='Sample', zorder=2)
for i in range(y.shape[1]-1):
    ax.plot(y[0,i:i+2], y[1,i:i+2], y[2,i:i+2], color='C2', linewidth=2)
y = ZZ0[:,int(20*ratio):int(120*ratio)+1]
ax.scatter(y[0,:], y[1,:], y[2,:], marker='.', color='C0', label='Delay', zorder=1)
for i in range(y.shape[1]-1):
    ax.plot(y[0,i:i+2], y[1,i:i+2], y[2,i:i+2], color='C0', linewidth=2)
y = ZZ0[:,int(120*ratio):int(136*ratio)]
ax.scatter(y[0,:], y[1,:], y[2,:], marker='.', color='C3', label='Decision', zorder=2)
for i in range(y.shape[1]-1):
    ax.plot(y[0,i:i+2], y[1,i:i+2], y[2,i:i+2], color='C3', linewidth=2)

y = ZZ1[:,:int(20*ratio)+1]
ax.scatter(y[0,:], y[1,:], y[2,:], marker='.', color='C5', label='Sample', zorder=2)
for i in range(y.shape[1]-1):
    ax.plot(y[0,i:i+2], y[1,i:i+2], y[2,i:i+2], color='C5', linewidth=2)
y = ZZ1[:,int(20*ratio):int(120*ratio)+1]
ax.scatter(y[0,:], y[1,:], y[2,:], marker='.', color='C1', label='Delay', zorder=1)
for i in range(y.shape[1]-1):
    ax.plot(y[0,i:i+2], y[1,i:i+2], y[2,i:i+2], color='C1', linewidth=2)
y = ZZ1[:,int(120*ratio):int(136*ratio)]
ax.scatter(y[0,0], y[1,0], y[2,:], marker='.', color='C4', label='Decision', zorder=2)
for i in range(y.shape[1]-1):
    ax.plot(y[0,i:i+2], y[1,i:i+2], y[2,i:i+2], color='C4', linewidth=2)
plt.legend()
plt.show()
```
